```python
x = dataset.drop(columns = ['PitNextLap'])
y = dataset['PitNextLap']
```

```python
y = y.astype('int32')
```

```python
x
```

```python
one_hot_encode_cols = ['Compound', 'Race']
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
from sklearn.compose import ColumnTransformer
ct = ColumnTransformer([
    ('one_hot_endcoing', OneHotEncoder(), one_hot_encode_cols),
], remainder='passthrough')
```

```python
x = ct.fit_transform(x)
```

```python
cols = ct.get_feature_names_out()
x = pd.DataFrame(x, columns = (name.split('__')[1] for name in cols))
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, stratify=y)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from xgboost import XGBClassifier
```

```python
models = [
    ('Random Forest', RandomForestClassifier(n_estimators = 500)),
    ('Decision Tree', DecisionTreeClassifier()),
    ('Logistic Regression', LogisticRegression()),
    ('KNN', KNeighborsClassifier()),
    ('SVM', SVC()),
    ('XGBoost', XGBClassifier()),
]
```

```python
from sklearn.model_selection import KFold
from sklearn.model_selection import cross_val_score
```

```python
kf = KFold(n_splits = 5, shuffle = True, random_state = 42)
for (name, model) in models:
  scores = cross_val_score(model, x, y, cv = kf, scoring = 'accuracy')
  print(name, scores.mean(), scores.std())
```

```python
kf = KFold(n_splits = 5, shuffle = True, random_state = 42)
cross_val_score(RandomForestClassifier(n_estimators = 200), x.drop(columns = ['Year']), y, cv = kf, scoring = 'accuracy').mean()
```
