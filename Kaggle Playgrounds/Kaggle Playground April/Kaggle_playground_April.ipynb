```python
import tensorflow as tf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
dataset = pd.read_csv('/content/train.csv')
```

```python
dataset
```

```python
dataset.info()
```

```python
dataset.describe()
```

```python
dataset = dataset.drop(['id'], axis = 1)
```

```python
dataset
```

```python
for column in dataset.columns:
  if dataset[column].dtype == 'object':
    print(column, ' : ', dataset[column].unique() , '      Different values: ' , len(dataset[column].unique()))
```

```python
dataset['Mulching_Used'] = dataset['Mulching_Used'].map({'No':0, 'Yes':1})
dataset['Irrigation_Need'] = dataset['Irrigation_Need'].map({'Low': 0, 'Medium':1, 'High':2})
```

```python
x = dataset.drop(['Irrigation_Need'], axis = 1)
y = dataset['Irrigation_Need']
```

```python
y
```

```python
categorical_cols = []
for column in dataset.columns:
  if dataset[column].dtype == 'object':
    categorical_cols.append(column)
```

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

ct = ColumnTransformer(
    transformers=[
        ('onehotencoder', OneHotEncoder(drop='first', sparse_output=False), categorical_cols),
    ]
    , remainder='passthrough'
)
```

```python
encoded_values = ct.fit_transform(x)
encoded_columns = ct.get_feature_names_out()
encoded_dataset = pd.DataFrame(encoded_values, columns=encoded_columns)
```

```python
encoded_dataset
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(encoded_dataset, y, test_size=0.2, random_state=42, stratify=y)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier(n_estimators=200)
model.fit(xtrain, ytrain)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(model.predict(xtest), ytest)
```

```python
model_tf = tf.keras.models.Sequential([
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(10, activation='relu'),
    tf.keras.layers.Dense(3, activation='softmax')
])

model_tf.compile(optimizer='adam',
                 loss='sparse_categorical_crossentropy',
                 metrics=['accuracy'])
```

```python
Final_model = RandomForestClassifier(n_estimators=200)
Final_model.fit(encoded_dataset, y)
```

```python
test_dataset = pd.read_csv('/content/test.csv')
id = test_dataset['id']
test_dataset.drop('id', axis = 1)
test_dataset['Mulching_Used'] = test_dataset['Mulching_Used'].map({'No':0, 'Yes':1})
test_dataset = pd.DataFrame(ct.transform(test_dataset), columns=encoded_columns)
```

```python
predicted_values = Final_model.predict(test_dataset)
```
