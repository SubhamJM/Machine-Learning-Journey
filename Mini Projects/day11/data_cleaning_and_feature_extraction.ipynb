```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv('/content/train_FD001.csv')
```

```python
dataset
```

```python
dataset = dataset.drop(['Unnamed: 0.1','Unnamed: 0'], axis = 1)
```

```python
fig, ax = plt.subplots(figsize=(25, 20))
sns.heatmap(dataset.corr(), annot = True, cbar = True, cmap = 'coolwarm', fmt=".2%")
```

```python
dataset.columns
```

```python
cols_to_drop = ['op_setting_3','sensor_1','sensor_5','sensor_6','sensor_10','sensor_16','sensor_18','sensor_19']
```

```python
dataset = dataset.drop(cols_to_drop, axis = 1)
```

```python
dataset
```

```python
sensor_list = ['sensor_2', 'sensor_3','sensor_4','sensor_7','sensor_8','sensor_9','sensor_11','sensor_12','sensor_13','sensor_14','sensor_15','sensor_17','sensor_20','sensor_21']
```

```python
for sensor in sensor_list:
  sns.relplot(dataset[dataset['engine_id'] == 1], x = sensor, y = 'RUL', kind = 'line', row = 5)
```

```python
sensors_to_smooth = ['sensor_2', 'sensor_3', 'sensor_4', 'sensor_7','sensor_9', 'sensor_11', 'sensor_12','sensor_17','sensor_21','sensor_15','sensor_14']
```

```python
dataset = dataset.reset_index(drop=True)
dataset[sensors_to_smooth] = dataset.groupby('engine_id')[sensors_to_smooth].transform(
    lambda x : x.rolling(window=10, min_periods=1).mean()
)
```

```python
dataset
```

```python
fig, ax = plt.subplots(figsize=(25, 20))
sns.heatmap(dataset.corr(), annot = True, cbar = True, cmap = 'coolwarm', fmt=".2%")
```
