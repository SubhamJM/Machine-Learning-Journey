```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
```

```python
dataset = pd.read_csv('/content/insurance.csv')
```

```python
dataset
```

```python
insurance = pd.get_dummies(dataset, dtype = 'int16')
```

```python
insurance
```

```python
x = insurance.drop('charges', axis = 1)
y = insurance['charges']
```

```python
from sklearn.model_selection import train_test_split
xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size = 0.2, random_state = 42)
```

```python
tf.random.set_seed(42)

insurance_model = tf.keras.Sequential([
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

insurance_model.compile(loss = tf.keras.losses.mae,
                        optimizer = tf.keras.optimizers.SGD(),
                        metrics=['mae'])

insurance_model.fit(xtrain, ytrain, epochs=100)
```

```python
insurance_model.evaluate(xtest, ytest)
```

```python
tf.random.set_seed(42)

insurance_model = tf.keras.Sequential([
    tf.keras.layers.Dense(100),
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

insurance_model.compile(loss = tf.keras.losses.mae,
                        optimizer = tf.keras.optimizers.Adam(learning_rate=0.001),
                        metrics=['mae'])

insurance_model.fit(xtrain, ytrain, epochs=100)
```

```python
insurance_model.evaluate(xtest, ytest)
```

```python
tf.random.set_seed(42)

insurance_model = tf.keras.Sequential([
    tf.keras.layers.Dense(100),
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

insurance_model.compile(loss = tf.keras.losses.mae,
                        optimizer = tf.keras.optimizers.Adam(),
                        metrics=['mae'])

history = insurance_model.fit(xtrain, ytrain, epochs=500)
```

```python
insurance_model.evaluate(xtest, ytest)
```

```python
pd.DataFrame(history.history).plot()
plt.ylabel('loss')
plt.xlabel('epochs')
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
insurance = pd.read_csv('/content/insurance.csv')
```

```python
insurance
```

```python
from sklearn.compose import make_column_transformer
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
ct = make_column_transformer(
    (MinMaxScaler(), ['age','bmi','children']),
    (OneHotEncoder(handle_unknown='ignore'), ['sex', 'smoker', 'region'])
)

x = insurance.drop('charges', axis = 1)
y = insurance['charges']

xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, random_state = 42)

ct.fit(xtrain)

xtrain_normal = ct.transform(xtrain)
xtest_normal = ct.transform(xtest)
```

```python
xtrain_normal[0]
```

```python
xtrain.shape, xtrain_normal.shape
```

```python
import tensorflow as tf
```

```python
insurance_model_2 = tf.keras.Sequential([
    tf.keras.layers.Dense(100),
    tf.keras.layers.Dense(10),
    tf.keras.layers.Dense(1)
])

insurance_model_2.compile(loss=tf.keras.losses.mae,
                          optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
                          metrics=['mae'])

insurance_model_2.fit(xtrain_normal, ytrain, epochs=200)
```

```python
insurance_model_2.evaluate(xtest_normal, ytest)
```
