```python
#create a rank 2 tensor
rank_2_tensor = tf.constant([[10,2],[14,7]])
```

```python
rank_2_tensor.shape, rank_2_tensor.ndim
```

```python
rank_2_tensor
```

```python
#adding extra dimension to rank2 tensor
rank_3_tensor = rank_2_tensor[..., tf.newaxis]
```

```python
rank_3_tensor
```

```python
#alternative
tf.expand_dims(rank_2_tensor, axis = -1) #-1 means expand the ending col
```

```python
#alternative
tf.expand_dims(rank_2_tensor, axis = 0)
```
