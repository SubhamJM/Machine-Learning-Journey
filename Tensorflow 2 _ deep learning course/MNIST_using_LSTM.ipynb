```python
import tensorflow as tf

fashion_mnist = tf.keras.datasets.fashion_mnist
(x_train, y_train), (x_test, y_test) = fashion_mnist.load_data()

x_train, x_test = x_train / 255.0, x_test / 255.0

print(f"Training images shape: {x_train.shape}")
print(f"Training labels shape: {y_train.shape}")
```

```python
y_train
```

```python
T = x_train.shape[1]
D = x_train.shape[2]
C = x_train.shape[3] # Add channel dimension
N = x_train.shape[0]
```

```python
from tensorflow.keras.layers import Input, LSTM, GlobalMaxPool1D, Dense, Conv2D, MaxPool2D, Flatten
from tensorflow.keras.models import Model
```

```python
i = Input(shape=(T, D))
x = LSTM(128)(i)
x = Dense(64, activation='relu')(x)
x = Dense(10, activation='softmax')(x)

model = Model(i, x)
```

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(x_train, y_train, epochs=10, validation_data=(x_test, y_test))
```

```python
cnn_i = Input(shape=(T, D,1))
cnn_x = Conv2D(32, kernel_size=(3, 3), strides=(2,2), activation='relu')(cnn_i)
cnn_x = MaxPool2D(pool_size=(2, 2))(cnn_x)
cnn_x = Conv2D(64, kernel_size=(3, 3), strides=(2,2), activation='relu')(cnn_x)
cnn_x = MaxPool2D(pool_size=(2, 2))(cnn_x)
# Removed the third Conv2D and MaxPool2D layers to prevent spatial dimensions from becoming too small
# cnn_x = Conv2D(128, kernel_size=(3, 3), strides=(2,2), activation='relu')(cnn_x)
# cnn_x = MaxPool2D(pool_size=(2, 2))(cnn_x)
cnn_x = Flatten()(cnn_x)
cnn_x = Dense(10, activation='softmax')(cnn_x)

cnn_model = Model(cnn_i, cnn_x)
```

```python
cnn_model.compile(optimizer=tf.keras.optimizers.Adam(learning_rate=0.01), loss='sparse_categorical_crossentropy', metrics=['accuracy'])
cnn_model.fit(x_train, y_train, batch_size=32, epochs=10, validation_data=(x_test, y_test))
```
