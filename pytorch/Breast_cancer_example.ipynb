```python
import numpy as np
import pandas as pd
import torch
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv")
```

```python
df
```

```python
df.drop(['id', 'Unnamed: 32'], inplace = True, axis = 1)
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

xtrain, xtest, ytrain, ytest = train_test_split(df.drop(['diagnosis'], axis = 1), df['diagnosis'], test_size=0.2)
```

```python
ytrain
```

```python
scalar = StandardScaler()
xtrain = scalar.fit_transform(xtrain)
xtest = scalar.transform(xtest)
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
ytrain = encoder.fit_transform(ytrain)
ytest = encoder.transform(ytest)
```

```python
ytrain
```

```python
xtrain_tensor = torch.from_numpy(xtrain)
xtest_tensor = torch.from_numpy(xtest)
ytrain_tensor = torch.from_numpy(ytrain)
ytest_tensor = torch.from_numpy(ytest)
```

```python
xtrain_tensor
```

```python
learning_rate = 0.1
epochs = 1000
```

```python
xtrain_tensor.shape
```

```python
class SimpleNN():
  def __init__(self, x):
    self.weights = torch.rand(x.shape[1], 1, dtype=torch.float64, requires_grad=True) ## creates a 30x1 matrix
    self.bias = torch.zeros(1, dtype=torch.float64, requires_grad=True)

  def forward(self, x):
    z = torch.matmul(x, self.weights) + self.bias
    return torch.sigmoid(z)

  def binary_cross_entropy(self, y_pred, y):
    EPSILON  = 1e-10
    y_pred = torch.clamp(y_pred, EPSILON, 1-EPSILON)

    loss = -torch.mean(y * torch.log(y_pred) + (1 - y) * torch.log(1 - y_pred))
    return loss
```

```python
# creating model
model = SimpleNN(xtrain_tensor)
model.weights
```

```python
model.bias
```

```python
# define loop
for epoch in range(epochs):
  # forward pass
  y_pred = model.forward(xtrain_tensor)

  # loss calculate
  loss = model.binary_cross_entropy(y_pred, ytrain_tensor)
  print(f'epoch: {epoch+1}, loss: {loss.item()}')
  # backward pass
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  #zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()
```

```python
#model evaluation
with torch.no_grad():
  y_pred = model.forward(xtest_tensor)
  y_pred = (y_pred > 0.5).float()
```

```python
accuracy = (y_pred == ytest_tensor).float().mean()
print(accuracy)
```
