# **KEY FINDINGS:**
#1. The relationship between total bill and tip is almost linear suggesting that more bill will imply more tip by the customer

#2. There are more smoker customers than non smokers but the tip amount is not dependent on a person being smoker or not

#3. There are more male customers than female customers and the male customers usually give more tip than female customers

#4. On Weekends customers only come at Dinner time and in weekdays, they only come at lunch time

#5 The customer tip increases with their size linearly

#6 The most common total bill from customers is around 15-20 dollars and the most common tipping amount is around 2-4 dollars

```python
dataset
```

```python
sns.heatmap(dataset.corr(numeric_only=True), annot = True, cbar = True)
```

```python
sns.relplot(dataset, x = 'total_bill', y = 'tip')
```

```python
sns.relplot(dataset, x = 'total_bill', y = 'tip', hue = 'sex', col = 'day', col_wrap=2, height = 4)
```

# **In all the days from thursday to sundary, we see a almost linear increase in the amount of Tip compared to the total bill of a customer regardless of them being a male or female**

```python
sns.relplot(dataset, x = 'total_bill', y = 'tip', col = 'sex')
```

# **We notice that a lot more men get more total bill than women and also pay higher tip compared to women at higher bills**

```python
sns.catplot(dataset, x = 'smoker', y = 'tip', kind='box')
```

# **We see that smokers and non smokers tip almost the same amount if we ignore the outliers**

```python
sns.displot(dataset, x = 'sex', hue = 'smoker', multiple = 'dodge', shrink = 0.8)
```

# **We can see that a lot more males come to the shop compared to females, and most of the customers are smokers in both male and female category**

```python
sns.displot(dataset, x = 'sex', col = 'time', row = 'day', hue = 'smoker',multiple = 'dodge', shrink = 0.8, height = 3)
```

# **We notice that in the weekends no customers come at lunch time and only come at dinner time, while in strict weekdays, most of the customers come only at lunch time.**

```python
sns.boxplot(dataset, x = 'size', y = 'tip')
```

# **We see that generally more size will imply more tip from the customer**

```python
sns.displot(dataset, x = 'total_bill', col = 'day', col_wrap=2, height = 3)
```

# **We observe that most of the days the most common total_bill is at around 15-20 dollars**

```python
sns.displot(dataset, x = 'tip', col = 'day', col_wrap=2, height = 3)
```

# **The Most common tip is also around 2-4 dollars on each day**
