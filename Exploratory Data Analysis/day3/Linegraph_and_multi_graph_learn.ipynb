```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style='darkgrid')
```

```python
dataset = sns.load_dataset('dowjones')
```

```python
dataset
```

```python
sns.relplot(data=dataset, x = 'Date', y = 'Price', kind = 'line')
```

```python
fmri = sns.load_dataset('fmri')
fmri
```

```python
sns.relplot(data = fmri, x='timepoint', y ='signal', kind='line')
```

```python
sns.lineplot(data=fmri, x = 'timepoint', y = 'signal', errorbar=None)
```

```python
sns.relplot(data=fmri, x = 'timepoint', y = 'signal', hue = 'event', kind = 'line', errorbar=None)
```

```python
sns.relplot(data=fmri, x = 'timepoint', y = 'signal', hue = 'event', style='region', kind = 'line', errorbar=None)
```

```python
sns.lineplot(data=fmri.query("event=='stim'"), x = 'timepoint', y ='signal', units='subject', estimator=None, hue = 'region')
```

# **Multi graph learnng**

```python
tips = sns.load_dataset('tips')
sns.relplot(data=tips, x = 'total_bill', y = 'tip', hue = 'smoker', col = 'time')
```

```python
sns.relplot(data=fmri, x = 'timepoint', y = 'signal', hue = 'subject', col = 'event', row = 'region', kind = 'line')
```

```python
sns.relplot(data = fmri.query('region=="frontal"'), kind = 'line', x = 'timepoint', y ='signal', hue = 'event', col = 'subject', col_wrap=4, height =3, linewidth=2.5)
```
