```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
sns.set_theme(style='darkgrid')
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.catplot(tips, x = 'smoker', y = 'tip', hue = 'sex')
```

```python
sns.catplot(tips, x = 'sex', y = 'tip', kind = 'swarm')
```

```python
sns.catplot(tips, x = 'smoker', y = 'tip', order = ['No', 'Yes'])
```

```python
sns.catplot(tips, x = 'day', y = 'tip', kind = 'box', hue = 'smoker')
```

```python
tips['weekend'] = tips['day'].isin(['Sat', 'Sun'])
```

```python
sns.catplot(tips, x = 'day', y = 'tip', hue = 'weekend', kind = 'box')
```

```python
diamonds = sns.load_dataset('diamonds')
diamonds
```

```python
sns.catplot(diamonds, x = 'color', y = 'price', kind = 'boxen')
```

```python
sns.catplot(tips, x = 'day', y = 'tip', hue = 'weekend', kind = 'violin')
```

```python
sns.catplot(tips, x = 'day', y = 'tip', hue = 'sex', split = True, kind = 'violin')
```

```python
sns.catplot(tips, x = 'day', y = 'tip', hue = 'sex', kind = 'violin')
```
