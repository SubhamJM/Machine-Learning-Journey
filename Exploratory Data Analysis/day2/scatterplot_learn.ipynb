# **LEARNING SCATTER PLOT IN SEABORN**


```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
```

```python
sns.set_theme(style='darkgrid')
```

```python
dataset = sns.load_dataset('tips')
sns.relplot(data=dataset, x ='total_bill', y = 'tip')
```

```python
dataset.columns
```

```python
print(dataset.head(5))
print(dataset.shape)
```

```python
sns.relplot(data=dataset, x='total_bill', y ='tip', hue='sex')
```

```python
sns.relplot(data=dataset, x='total_bill', y = 'tip', style = 'smoker', hue = 'sex')
```

```python
sns.relplot(kind= 'scatter', data=dataset, x = 'total_bill', y = 'tip', hue = 'size',palette="ch:r=-.5,l=.75")
```

```python
sns.relplot(kind='scatter', data=dataset, x = 'total_bill', y = 'tip', size = 'size')
```
